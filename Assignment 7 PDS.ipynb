**Theory**

1.What is the difference between Binomial, Poisson, and Normal distributions?

The Binomial, Poisson, and Normal distributions are key ideas in Statistics, and each is applied in different situations depending on the type of data. The Binomial distribution is used when there is a fixed number of repeated experiments, where each one results in either success or failure, and the probability of success remains the same throughout. For example, finding how many times a coin lands on heads in several tosses follows this distribution.

On the other hand, the Poisson distribution is used to represent how often an event occurs within a specific period of time or space, assuming the events happen independently and at a steady average rate, such as counting the number of calls received in an hour. The Normal distribution, in contrast, is a continuous distribution that shows how values are spread around a central mean in a symmetric, bell-shaped pattern, and is commonly used to model natural data like heights or exam marks.

In short, the Binomial distribution focuses on counting successes in a fixed number of trials, the Poisson distribution measures how frequently events occur over an interval, and the Normal distribution explains how continuous data is distributed around an average value.


2.What is the real-world significance of the Poisson distribution?

The Poisson distribution is very important in real-world situations because it is used to estimate how often an event happens within a fixed time or space, especially when those events occur randomly and independently. In Statistics, it is particularly helpful for analyzing events that are uncertain but still occur at an average rate.

In practical life, it is widely applied in different fields. For example, in telecommunications, it helps estimate the number of calls or messages received in a given time so systems can be managed efficiently. In healthcare, hospitals use it to predict patient arrivals in emergency units and plan staffing accordingly. It is also useful in traffic management to estimate how many vehicles pass through a road or intersection, which supports better planning and design. In manufacturing, it is used to monitor product quality by estimating the number of defects in a production batch.

Overall, the Poisson distribution is valuable because it turns random occurrences into meaningful estimates, helping organizations make better decisions, allocate resources effectively, and plan more efficiently across various sectors.


3.What is the Central Limit Theorem and how does it relate to Normal distribution?

The central limit theorem is an important principle in Statistics that describes how sample averages behave. It states that when we take many random samples from a population and compute their means, the distribution of those means gradually becomes approximately normal as the sample size increases, no matter what the original population distribution looks like.

This concept is strongly connected to the normal distribution because it explains why the bell-shaped curve is so common in real-world data. Even if the original data is uneven or skewed, the distribution of sample means will still tend to form a normal shape when the sample size is sufficiently large. As a result, the normal distribution can be widely used in statistical techniques such as estimating population values and performing hypothesis testing, even when the original data is not perfectly normally distributed.


4.Why is Normal distribution widely used in machine learning and statistics?

The normal distribution is commonly used in machine learning and Statistics because it frequently occurs in real-world data and makes analysis easier. Many natural and practical phenomena, such as measurement errors, human heights, and random noise in datasets, tend to follow a symmetric, bell-shaped pattern. This is mainly explained by the central limit theorem, which shows that averages of random variables tend to form a normal distribution, making it a useful approximation in many situations.

Another reason for its widespread use is its simplicity in mathematical handling. Since it is defined by just the mean and standard deviation, it becomes easier to study, interpret, and apply in calculations. Many machine learning methods, including linear regression, logistic regression, and Gaussian-based models, assume normally distributed errors because this simplifies computations and improves efficiency.

It also plays a key role in statistical inference, helping in estimating unknown values, constructing confidence intervals, and performing hypothesis testing. In machine learning, it is applied in models like Gaussian Naive Bayes and in processes such as noise handling and feature scaling. Overall, its practical relevance and ease of use make the normal distribution one of the most important tools in both statistics and machine learning.


5.What is skewness in distributions and which distributions exhibit it?

Skewness is a concept in Statistics that describes how uneven or asymmetric a distribution is. When the data is not evenly balanced on both sides, the distribution is said to be skewed, meaning one side extends further than the other.

There are two types of skewness. In positive (right) skewness, the tail on the right side is longer, and most of the data points lie on the left. In negative (left) skewness, the tail on the left side is longer, and most values are clustered on the right side.

Different distributions display skewness differently. The normal distribution is completely symmetric and therefore has no skewness. The Poisson distribution generally shows positive skewness, especially when its mean is low. The binomial distribution can be skewed either to the left or right depending on the probability of success, but it becomes nearly symmetric when that probability is close to 0.5. Understanding skewness is important because it helps identify the direction and extent of imbalance in a dataset.



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm

# Set up a figure with subplots
fig, axs = plt.subplots(3, 3, figsize=(15, 12))
fig.suptitle("Real-World Analysis of Probability Distributions", fontsize=16)

# ------------------- BINOMIAL DISTRIBUTION -------------------
# Scenario: Number of defective items in a batch (defect probability = 0.1)
n, p = 50, 0.1
x_binom = np.arange(0, n+1)
pmf_binom = binom.pmf(x_binom, n, p)
cdf_binom = binom.cdf(x_binom, n, p)
samples_binom = binom.rvs(n, p, size=1000)

# PMF
axs[0, 0].bar(x_binom, pmf_binom, color='skyblue')
axs[0, 0].set_title("Defective Items in Batch (n=50, p=0.1)")
# CDF
axs[0, 1].step(x_binom, cdf_binom, where='mid', color='blue')
axs[0, 1].set_title("Binomial CDF")
# Histogram of Samples
axs[0, 2].hist(samples_binom, bins=np.arange(n+2)-0.5, density=True, color='navy', alpha=0.7)
axs[0, 2].set_title("Sample Distribution")

# ------------------- POISSON DISTRIBUTION -------------------
# Scenario: Number of website visits per minute
mu = 25
x_poisson = np.arange(0, 50)
pmf_poisson = poisson.pmf(x_poisson, mu)
cdf_poisson = poisson.cdf(x_poisson, mu)
samples_poisson = poisson.rvs(mu, size=1000)

# PMF
axs[1, 0].bar(x_poisson, pmf_poisson, color='lightgreen')
axs[1, 0].set_title("Website Visits per Minute (μ=25)")
# CDF
axs[1, 1].step(x_poisson, cdf_poisson, where='mid', color='green')
axs[1, 1].set_title("Poisson CDF")
# Histogram of Samples
axs[1, 2].hist(samples_poisson, bins=np.arange(51)-0.5, density=True, color='darkgreen', alpha=0.7)
axs[1, 2].set_title("Sample Distribution")

# ------------------- NORMAL DISTRIBUTION -------------------
# Scenario: Daily temperatures in a city (mean = 28°C, std deviation = 5°C)
mu_norm, sigma = 28, 5
x_norm = np.linspace(10, 45, 1000)
pdf_norm = norm.pdf(x_norm, mu_norm, sigma)
cdf_norm = norm.cdf(x_norm, mu_norm, sigma)
samples_norm = norm.rvs(mu_norm, sigma, size=1000)

# PDF
axs[2, 0].plot(x_norm, pdf_norm, color='salmon')
axs[2, 0].set_title("Daily Temperature Distribution (μ=28, σ=5)")
# CDF
axs[2, 1].plot(x_norm, cdf_norm, color='red')
axs[2, 1].set_title("Normal CDF")
# Histogram of Samples
axs[2, 2].hist(samples_norm, bins=30, density=True, color='darkred', alpha=0.7)
axs[2, 2].set_title("Sample Distribution")

# Layout and show
for row in axs:
    for ax in row:
        ax.grid(True)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()

# ------------------- BASIC STATISTICAL ANALYSIS -------------------
print("Binomial Distribution (Defective Items):")
print(f"Mean = {np.mean(samples_binom):.2f}, Variance = {np.var(samples_binom):.2f}")

print("\nPoisson Distribution (Website Visits):")
print(f"Mean = {np.mean(samples_poisson):.2f}, Variance = {np.var(samples_poisson):.2f}")

print("\nNormal Distribution (Daily Temperature):")
print(f"Mean = {np.mean(samples_norm):.2f}, Standard Deviation = {np.std(samples_norm):.2f}")
```
